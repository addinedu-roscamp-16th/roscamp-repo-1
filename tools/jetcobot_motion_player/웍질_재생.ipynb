## 3. 실제 로봇

⚠️ **웍 로봇(로봇2)은 세로 꽂기 웍 공구가 손목에 달려 있어야 한다.** 그리퍼 상태로 돌리면 화구·테이블에 부딪힐 수 있다.

- 로봇에 연결된 PC(라즈베리파이)에서 실행. 포트가 다르면 `connect("/dev/ttyUSB0")`
- 로봇 없이 흐름만 보려면 `connect(fake=True)`

```python
connect()            # 가짜 로봇: connect(fake=True)
```

```python
play(1, rate=0.5)    # 처음엔 절반 속도로. 괜찮으면 rate=1.0
```

```python
play(2, rate=0.5)
```

```python
play_all([1, 2, 3], rate=1.0)    # 이어서 전부
```

```python
stop()   # 비상 정지
```

## 5. 내 코드에서 직접 쓰기

`load()` 는 `(정보, [[시간초, J1..J6 도], ...])` 를 돌려준다. 직접 만든 각도 목록은 `save()` 로 같은 형식으로 저장.

```python
info, pts = load(2)
print(info["name"], len(pts), "점, 첫 점:", pts[0])
# save("내 동작", [[0.0, 0,0,0,0,0,0], [2.0, 10,0,0,0,0,0]], robot="로봇2", made_by="직접 계산", notes="")
```
